```python
#function to evaluate the polynomial function
def my_f(x):
    return x**5/10 + x**3 - 10*x**2 + 4*x + 7
#function to perform the bisection algorithm
def compute_root(f,x0,x1,tol,max_it):
    #ensure initial error > tol for loop to run and initialise counter to 0
    error = tol + 1.0
    counter = 0
    #iterates for as long as abs(f_mid) < 1.0e-6 and ensures the max number of iterations is not succeeded to guard against infinate loop error
    while error>tol and counter<max_it :
        # computes midpoint
        x_mid = (x0 + x1) / 2
        # evaluates function for first root on LHS and then for the midpoint
        f_x0 = f(x0)
        f_mid = f(x_mid)
        # checks between which roots f changes sign to assign x0 or x1 to xmid
        if f_mid * f_x0 <0:
            x1 = x_mid
            counter+=1
        else:
            x0 = x_mid
            counter+=1
        # assigns new error as f_mid to ensure abs(f_mid) < tol
        error = abs(f_mid)
        num_of_it = counter
    return x_mid, f_mid, num_of_it


print(compute_root(my_f,0,2,1.0e-6,1000))

def my_f2(x): #evaluate a 2nd polynomial function
    return x**5/9 + x**4 + x**2 - 8 *x**2 + 3*x + 7

print( compute_root(my_f2,0,2,1.0e-6,1000))

##tests##
x, f, num_it = compute_root(my_f, x0=0, x1=2, tol=1.0e-6, max_it=1000)
#test solution for function in exercise 2
assert math.isclose(x, 1.1568354368209839)

x2, f2, num_it2 = compute_root(my_f2, x0=0, x1=2, tol=1.0e-6, max_it=1000)
assert math.isclose(x2, 2.0)
    
```

```python
#import random module to access randint function
import random
#function to roll dice with n sides and return a number
def dice_roll(n):
    return random.randint(1,n)
#function to test if the dice roll function was fair
def test_fairness():
    #creates a list for each number on the dice all initially indexed at 0
    counts = [0] * 7
    #loop to roll dice of 6 sides 600 times
    for i in range (600):
        result = dice_roll(6)
        counts[result]+=1 #each time a number on the dice is landed on, the list is incremented by 1
# checks each value for number of rolls in the list and checks whether it was landed on a fair amount
    for num in range(1,7):
        if counts[num] > 150 or counts[num] < 50:
            return 'Dice is unfair'
    
    return 'Dice is fair'

print(test_fairness())
        
        
        
```

```python
#function to compute the area of a triangle by entering 3 vertices in a list
def area_of_triangle(vertices):
    #ensures they have entered 3 vertices
    if len(vertices) != 3:
        print('you must enter 3 vertices')
        return None
    #ensures 2d coordintes are entered
    for coord in vertices:
        if len(coord) != 2:
            print('enter a 2d co ordinate')
            return None
    #ensure each value entered is either an integer or a float
    for i in range (len(coord)):
        if type(coord[i]) != float and type(coord[i]) != int:
            print('invalid entry')
            return None
    #compute area
    (x0,y0), (x1,y1), (x2,y2) = vertices
    area = abs((x0*(y1 - y2) + x1*(y2-y0) + x2*(y0-y1))/2)
    return area

print (area_of_triangle([(0.0,0.0),(0.0,2.0),(3.0,0.0)]))
    
    
    
```

```python
#function to calcuate the area of a polygon of n vertices
def polygon_area(vertices,n):
    #set initial sum to 0
    sum = 0.0
    for i in range (n):
        #ensures vertices are ordered correctly as you move around the polygon
        ii = (i + 1) % n
        (xi,yi) = vertices[i]
        (x_ii,y_ii) = vertices[ii]
        sum += (xi*y_ii - x_ii*yi)
    area = abs(sum/2)
    return area

vertices0 = [(2,3), (7,3), (7,8), (2,8)]
print(polygon_area(vertices0,4))

##tests##
A = polygon_area(vertices0,4)
assert A == 25.0
    
```

```python
#function which adds together 2 vectors of arbitary length
def sum_vector(x,y):
    #Ensures vector sizes match
    if len(x) == len(y):
        print ('Valid')
    else:
        print ('Warning - cannot carry out addition')
        return None
    new_vector = []
    for i in range (len(x)):
        #adds each index in the vectors and stores result in new vector
        new_vector.append(x[i] + y[i])
    return new_vector

##tests##
x0 = [0, 4.3, -5, 7]
y0 = [-2, 7, -15, 1]  
result = sum_vector(x0,y0)
print(result)
assert result == [-2, 11.3, -20, 8]
    
```
